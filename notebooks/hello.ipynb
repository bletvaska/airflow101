## Lab 2: Minimalna hodnota

Vytvorte funkciu `min()`. Táto funkcia dostane ako parameter zoznam čísel. Funkcia vráti najmensie (minimalne) cislo z tohto zoznamu.

Priklad pouzitia:

```python
>>> min( [1,2,3,4,5] )
1
>>> min( [50, 40, 30] )
30
>>> min( [1,1,1,1,1] )
1
```


````python
def min(numbers: list) -> float:
    numbers.sort()
    return numbers[0]

min( [1,2,3,4,5] )
````

````python
def min(numbers: list) -> float:
    numbers.sort()
    #dlzka = len(numbers)
    mincislo = numbers[0]
   
    print('Minimalne cislo je :', mincislo)
       
    return mincislo

min( [1,2,3,4,5] )
````

````python
def min(zoznam : list) -> float:
    najmensie = zoznam[0]


    for cislo in zoznam:
        if cislo < najmensie:
            najmensie = cislo


    return najmensie

min( [1,2,3,4,5] )
````
